```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month_name()
```

```python
grouped_orders = df.groupby(["order_year", "order_month"])["order_id"].size().reset_index(name = "order_count")
pivot_data = grouped_orders.pivot(index="order_month", columns = "order_year", values="order_count")

months_order = [
    "January", "February", "March", "April", "May", "June", 
    "July", "August", "September", "October", "November", "December"
]
pivot_data = pivot_data.reindex(months_order)

pivot_data
```
